## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# 1. Make a working copy so the raw data stays untouched
df_clean = df.copy()

# 2. Drop columns that are almost entirely missing and add little value
#    connected_handling_time is 99.7% missing -> not usable as a feature
df_clean.drop(columns=['connected_handling_time'], inplace=True)

# 3. Convert date/time columns to proper datetime objects
df_clean['order_date_time'] = pd.to_datetime(df_clean['order_date_time'], format='%d/%m/%Y %H:%M', errors='coerce')
df_clean['Issue_reported at'] = pd.to_datetime(df_clean['Issue_reported at'], format='%d/%m/%Y %H:%M', errors='coerce')
df_clean['issue_responded'] = pd.to_datetime(df_clean['issue_responded'], format='%d/%m/%Y %H:%M', errors='coerce')
df_clean['Survey_response_Date'] = pd.to_datetime(df_clean['Survey_response_Date'], format='%d-%b-%y', errors='coerce')

# 4. Derive a response time feature (in minutes) from the two timestamp columns
#    This replaces connected_handling_time, which was almost entirely missing
df_clean['response_time_min'] = (df_clean['issue_responded'] - df_clean['Issue_reported at']).dt.total_seconds() / 60

# 5. Handle invalid response times (negative values = data entry errors, response logged before issue)
print(f"Negative response times found: {(df_clean['response_time_min'] < 0).sum()}")
df_clean.loc[df_clean['response_time_min'] < 0, 'response_time_min'] = pd.NA

# 6. Handle invalid Item_price (one record has price = 0, likely a data error or free item)
df_clean.loc[df_clean['Item_price'] <= 0, 'Item_price'] = pd.NA

# 7. Standardize text columns - strip whitespace, fix casing inconsistencies
text_cols = ['channel_name', 'category', 'Sub-category', 'Customer_City',
             'Product_category', 'Agent_name', 'Supervisor', 'Manager',
             'Tenure Bucket', 'Agent Shift']
for col in text_cols:
    df_clean[col] = df_clean[col].astype(str).str.strip()
    df_clean[col] = df_clean[col].replace('nan', pd.NA)  # restore actual NaNs after str conversion

# 8. Fix the known typo in Product_category
df_clean['Product_category'] = df_clean['Product_category'].replace('Home Appliences', 'Home Appliances')

# 9. Standardize Customer_City casing (currently a mix of cases)
df_clean['Customer_City'] = df_clean['Customer_City'].str.title()

# 10. Create a flag for whether the customer left a written remark
df_clean['has_remarks'] = df_clean['Customer Remarks'].notna().astype(int)

# 11. Create a flag for whether the ticket is tied to an order
df_clean['has_order'] = df_clean['Order_id'].notna().astype(int)

# 12. Extract day of week / hour from Issue_reported at - useful for EDA later
df_clean['issue_day_of_week'] = df_clean['Issue_reported at'].dt.day_name()
df_clean['issue_hour'] = df_clean['Issue_reported at'].dt.hour

# 13. Final check
print("Shape after wrangling:", df_clean.shape)
df_clean.info()
````

### What all manipulations have you done and insights you found?

1. **Dropped `connected_handling_time`** – this column was missing in 99.7% of rows,
   making it unusable as a feature.
2. **Converted date/time columns** (`order_date_time`, `Issue_reported at`,
   `issue_responded`, `Survey_response_Date`) from string to proper datetime format,
   so they could be used for time-based calculations and EDA.
3. **Derived a new feature `response_time_min`** (the difference between `issue_responded`
   and `Issue_reported at`, in minutes) to act as a usable substitute for the largely missing
   `connected_handling_time` column.
4. **Corrected invalid response times** – found 3,128 rows (~3.6%) where the response
   timestamp was *before* the reported timestamp, producing negative durations. These were
   treated as data entry errors and set to missing rather than left as invalid negative values.
5. **Corrected an invalid `Item_price`** – one record had a price of 0, which was treated as
   missing rather than a real zero-cost item.
6. **Standardized text columns** – stripped extra whitespace and fixed inconsistent casing
   across categorical columns (e.g., `channel_name`, `category`, `Customer_City`) to avoid
   the same value being treated as multiple categories due to formatting differences.
7. **Fixed a typo** – `Home Appliences` was corrected to `Home Appliances` in
   `Product_category`.
8. **Standardized `Customer_City`** to title case, since city names appeared in a mix of
   cases (e.g., "HYDERABAD" vs "hyderabad").
9. **Created two flag features**:
   - `has_remarks` — whether the customer left written feedback (only ~33% did).
   - `has_order` — whether the ticket is linked to a specific order (~79% are).
10. **Extracted day-of-week and hour from `Issue_reported at`** to enable time-pattern
    analysis (e.g., whether certain hours/days see lower satisfaction).

## Insights Found

- **Response time data is unreliable in places**: ~3.6% of tickets show a response
  timestamp earlier than the issue timestamp, pointing to either system logging errors or
  timezone/clock mismatches between systems. This needs to be kept in mind when interpreting
  handling-time-related conclusions later.
- **Most tickets have no written customer remark** (67% missing) — satisfaction analysis
  will rely heavily on the numeric CSAT score rather than text feedback for the majority
  of records.
- **Order-related details (price, product category, order date) are missing for roughly
  80% of tickets** — likely because many tickets (e.g., general queries, feedback,
  cancellations) simply aren't tied to a specific order. This isn't necessarily "bad"
  missing data — it may be structurally missing based on ticket type, which is worth
  checking in EDA (e.g., does `category` predict whether `Order_id` exists?).
- **The data is clean at the structural level** — no duplicate rows or duplicate ticket IDs,
  so no risk of double-counting interactions.

#### Chart - 1

CSAT Score Distribution

````python
plt.figure(figsize=(8,5))
ax = sns.countplot(x='CSAT Score', data=df, order=[1,2,3,4,5], palette='viridis')

# add percentage labels on top of bars
total = len(df)
for p in ax.patches:
    pct = f'{100 * p.get_height() / total:.1f}%'
    ax.annotate(pct, (p.get_x() + p.get_width()/2, p.get_height()),
                ha='center', va='bottom', fontsize=10)

plt.title('Distribution of CSAT Scores', fontsize=14)
plt.xlabel('CSAT Score (1 = Very Dissatisfied, 5 = Very Satisfied)')
plt.ylabel('Number of Tickets')
plt.tight_layout()
plt.show()
````

##### 1. Why did you pick the specific chart?

A count plot (bar chart) is the right choice here because CSAT Score is the target
variable for this whole project, and it's a discrete, ordinal categorical variable
(1 to 5). A bar chart shows both the relative frequency of each score and lets you
compare category sizes directly — which is exactly what's needed before doing any
further analysis or modeling. This is also the natural first chart in any classification-
style project: you need to understand your target distribution before you can
meaningfully analyze what drives it.

##### 2. What is/are the insight(s) found from the chart?

- The CSAT scores are heavily skewed toward 5 — about 69.4% of all tickets received
  the highest score.
- Combined, scores of 4 and 5 ("satisfied") make up roughly 82.5% of all responses.
- Only about 2.98% of customers gave a neutral score of 3, and just 1.49% gave a 2 —
  these are by far the rarest classes.
- Interestingly, the distribution isn't a smooth taper from 5 down to 1. Score 1
  (13.07%) is almost as frequent as score 4 (13.06%), creating a bimodal-like shape:
  customers tend to be either very satisfied or very dissatisfied, with comparatively
  few landing in the middle (2 or 3).
- This means the dataset is highly imbalanced, with the minority classes (2 and 3)
  making up under 5% combined. This has direct implications for the Model Building
  stage — without resampling or class-weighting, a model could achieve high accuracy
  just by always predicting 5, while completely failing to identify dissatisfied
  customers, which is the group that actually matters most for business action.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

**Positive business impact:**
Knowing that ~82.5% of customers are already satisfied is a good baseline signal —
Flipkart's support function is performing reasonably well overall. This insight helps
set realistic expectations: the business goal isn't to "fix everything" but to focus
resources on the smaller, identifiable group of dissatisfied customers (scores 1-2,

**Insights that could signal risk to growth:**
The "polarized" shape of the distribution — almost as many 1s as 4s — is a warning sign.

distinct subgroup having a genuinely bad experience (warranting a 1, not a 3). If left
unaddressed, this group of ~13% giving the lowest possible score represents a real churn
and brand-reputation risk — dissatisfied customers are more likely to leave negative
reviews, switch to competitors, or stop using the platform's support channels (and
potentially the platform itself).

There's also a modeling risk this insight surfaces: because classes 2 and 3 are so rare
(under 5% combined), any model trained on this data without addressing the imbalance
will struggle to reliably detect "mildly dissatisfied" customers (the ones who might
still be retained with the right intervention) — it will effectively treat the problem
as "very happy vs. very unhappy" by default, which could cause the business to miss
early warning signs before a customer becomes a 1.

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

Answer Here.

### Hypothetical Statement - 1

 Does response time significantly affect customer satisfaction?

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H₀):
There is no significant difference in response time between satisfied customers
(CSAT Score 4-5) and dissatisfied customers (CSAT Score 1-2). Any observed
difference is due to random chance.

Alternate Hypothesis (H₁):
Satisfied customers (CSAT Score 4-5) have significantly lower response times
than dissatisfied customers (CSAT Score 1-2).

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
from scipy import stats

# Step 1: Split into two groups based on CSAT
satisfied    = df_clean[df_clean['CSAT Score'] >= 4]['response_time_min'].dropna()
dissatisfied = df_clean[df_clean['CSAT Score'] <= 2]['response_time_min'].dropna()

print(f"Satisfied group    → n: {len(satisfied)}, Mean: {satisfied.mean():.2f} min, Median: {satisfied.median():.2f} min")
print(f"Dissatisfied group → n: {len(dissatisfied)}, Mean: {dissatisfied.mean():.2f} min, Median: {dissatisfied.median():.2f} min")

# Step 2: Check normality (Shapiro-Wilk on a sample — full dataset too large)
s_sample = satisfied.sample(500, random_state=42)
d_sample = dissatisfied.sample(500, random_state=42)

_, p_norm_sat  = stats.shapiro(s_sample)
_, p_norm_dis  = stats.shapiro(d_sample)
print(f"\nShapiro-Wilk p-value (satisfied):    {p_norm_sat:.2e}")
print(f"Shapiro-Wilk p-value (dissatisfied): {p_norm_dis:.2e}")
print("→ Both groups are NOT normally distributed. Proceeding with non-parametric test.")

# Step 3: Mann-Whitney U Test
stat, p_value = stats.mannwhitneyu(satisfied, dissatisfied, alternative='less')
print(f"\nMann-Whitney U Statistic : {stat:.0f}")
print(f"P-Value                  : {p_value:.2e}")

alpha = 0.05
if p_value < alpha:
    print(f"\nResult: P-value ({p_value:.2e}) < α ({alpha})")
    print("→ Reject the Null Hypothesis.")
    print("→ Satisfied customers have significantly LOWER response times than dissatisfied customers.")
else:
    print(f"\nResult: P-value ({p_value:.2e}) >= α ({alpha})")
    print("→ Fail to Reject the Null Hypothesis.")
````

##### Which statistical test have you done to obtain P-Value?

Mann-Whitney U Test (also known as the Wilcoxon Rank-Sum Test) — a non-parametric
two-sample test used to determine whether one group tends to have significantly
higher or lower values than another group.

##### Why did you choose the specific statistical test?

Three reasons drove this choice:

1. Non-normality of data:
   Before selecting any test, normality was checked using the Shapiro-Wilk test on
   a 500-sample from each group. Both groups returned p-values near zero (9.25e-40
   and 2.21e-35), meaning the normality assumption is strongly violated. A
   parametric t-test (which assumes both groups are normally distributed) would
   therefore be inappropriate here.

2. Heavily skewed, right-tailed data:
   Response time in minutes is extremely right-skewed — a handful of very long
   delays (up to 5,758 minutes) pull the mean far above the median (e.g., satisfied
   mean is 138 min but median is just 5 min). The Mann-Whitney U test compares
   distributions based on rank ordering rather than raw values, making it robust
   to this kind of skew and to extreme outliers.

3. Large, independent samples:
   The two groups (satisfied: n=68,126, dissatisfied: n=12,175) are independent
   of each other — one customer's satisfaction score doesn't influence another's
   response time — which is the fundamental assumption of Mann-Whitney U. The test
   is also reliable and powerful with large samples even under non-normality,
   making it the ideal choice here over alternatives like a permutation test
   (computationally expensive at this scale) or a Kruskal-Wallis test (designed
   for more than 2 groups).

### Hypothetical Statement - 2

Does the support channel significantly affect customer satisfaction (CSAT Score)?

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Null Hypothesis (H₀):
There is no significant difference in CSAT Score across the three support channels
(Email, Inbound, Outcall). Any observed difference in average satisfaction is due
to random chance.

Alternate Hypothesis (H₁):
At least one support channel has a significantly different CSAT Score distribution
compared to the others — i.e., the channel through which a customer contacts support
has a significant effect on their satisfaction level.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
from scipy import stats

# Step 1: Separate CSAT scores by channel
email   = df[df['channel_name'] == 'Email']['CSAT Score']
inbound = df[df['channel_name'] == 'Inbound']['CSAT Score']
outcall = df[df['channel_name'] == 'Outcall']['CSAT Score']

print(f"Email   → n: {len(email)},   Mean CSAT: {email.mean():.3f}")
print(f"Inbound → n: {len(inbound)}, Mean CSAT: {inbound.mean():.3f}")
print(f"Outcall → n: {len(outcall)}, Mean CSAT: {outcall.mean():.3f}")

# Step 2: Check normality on a sample from each group
for name, grp in [('Email',   email.sample(500, random_state=42)),
                  ('Inbound', inbound.sample(500, random_state=42)),
                  ('Outcall', outcall.sample(500, random_state=42))]:
    _, p_norm = stats.shapiro(grp)
    print(f"Shapiro-Wilk p-value ({name}): {p_norm:.2e}")

print("→ All three groups are NOT normally distributed. Proceeding with non-parametric test.")

# Step 3: Kruskal-Wallis Test (non-parametric, 3 independent groups)
h_stat, p_value = stats.kruskal(email, inbound, outcall)
print(f"\nKruskal-Wallis H-Statistic : {h_stat:.4f}")
print(f"P-Value                    : {p_value:.4e}")

alpha = 0.05
if p_value < alpha:
    print(f"\nResult: P-value ({p_value:.4e}) < α ({alpha})")
    print("→ Reject the Null Hypothesis.")
    print("→ At least one channel has a significantly different CSAT distribution.")

# Step 4: Post-hoc Pairwise Mann-Whitney U with Bonferroni Correction
print("\n--- Post-Hoc Pairwise Comparisons (Bonferroni Corrected) ---")
pairs = [('Email', 'Inbound', email, inbound),
         ('Email', 'Outcall', email, outcall),
         ('Inbound', 'Outcall', inbound, outcall)]

bonferroni_alpha = alpha / len(pairs)
print(f"Bonferroni Corrected α: {bonferroni_alpha:.4f}\n")

for a, b, ga, gb in pairs:
    stat, p = stats.mannwhitneyu(ga, gb, alternative='two-sided')
    sig = '✓ Significant' if p < bonferroni_alpha else '✗ Not Significant'
    print(f"{a} vs {b}: U={stat:.0f}, p={p:.4e} → {sig}")
````

##### Which statistical test have you done to obtain P-Value?

Kruskal-Wallis H Test — a non-parametric one-way analysis of variance (ANOVA)
that tests whether three or more independent groups come from the same distribution.
Followed by post-hoc Pairwise Mann-Whitney U Tests with Bonferroni correction
to identify exactly which channel pairs drive the overall significant result.

##### Why did you choose the specific statistical test?

1. Three groups, not two:
   Since we are comparing CSAT across three channels (Email, Inbound, Outcall),
   a two-sample test like Mann-Whitney U is not directly applicable. Kruskal-Wallis
   is the natural extension of Mann-Whitney U to three or more groups, testing
   whether any group has a stochastically different distribution from the others.

2. Non-normality confirmed:
   Shapiro-Wilk tests on samples from all three groups returned p-values near zero
   (1.73e-31, 4.25e-33, 2.95e-33), strongly rejecting normality in every group.
   This rules out a one-way parametric ANOVA, which requires normality within groups.

3. Ordinal discrete target:
   CSAT Score takes only integer values 1-5, which makes it ordinal rather than
   truly continuous. Kruskal-Wallis works on ranks rather than raw values,
   making it well-suited to ordinal data where the spacing between scores (e.g.,
   the difference between a 4 and a 5) isn't guaranteed to be meaningful.

4. Bonferroni-corrected post-hoc testing:
   When Kruskal-Wallis rejects H₀, it only tells you that *some* difference
   exists — not which specific channels differ. Post-hoc pairwise Mann-Whitney
   U tests with Bonferroni correction (adjusted α = 0.0167 for 3 comparisons)
   were added to identify exactly where the difference lies, while controlling
   for the risk of false positives from running multiple comparisons. The result
   shows Email is significantly different from both Inbound and Outcall, but
   Inbound and Outcall are statistically indistinguishable from each other
   (p = 0.085 > 0.0167).

### Hypothetical Statement - 3

##### Why did you choose the specific statistical test?

### 9. Handling Imbalanced Dataset

##### Do you think the dataset is imbalanced? Explain Why.

Yes — the dataset is significantly imbalanced.

Class distribution of CSAT Score:
  - Score 5 (Very Satisfied): 59,617 samples → 69.4%
  - Score 1 (Very Dissatisfied): 11,230 samples → 13.1%
  - Score 4: 11,219 samples → 13.1%
  - Score 3: 2,558 samples → 3.0%
  - Score 2: 1,283 samples → 1.5%

The majority class (Score 5) outnumbers the smallest minority class (Score 2)

of the data. A naive model that always predicts 5 would achieve ~69% accuracy

a critical problem to address.

````python
# Handling Imbalanced Dataset
from imblearn.over_sampling import SMOTE

print("Class distribution BEFORE balancing:")
print(y_train.value_counts().sort_index())

# Apply SMOTE to training set only (never to test set)
smote = SMOTE(random_state=42, k_neighbors=5)
X_train_bal, y_train_bal = smote.fit_resample(X_train_scaled, y_train)

print("\nClass distribution AFTER SMOTE:")
print(pd.Series(y_train_bal).value_counts().sort_index())
print(f"\nTraining set size after SMOTE: {X_train_bal.shape[0]} rows")
````

##### What technique did you use to handle the imbalance dataset and why? (If needed to be balanced)

SMOTE (Synthetic Minority Over-sampling Technique) was applied exclusively to
the training set.

Why SMOTE over other techniques:

1. Over SMOTE alternatives (Random Oversampling):
   Random oversampling simply duplicates existing minority rows, which can cause
   the model to overfit to specific training examples. SMOTE generates new
   synthetic minority samples by interpolating between real minority instances
   and their k-nearest neighbors, creating more diverse and generalizable
   training examples.

2. Over Undersampling:
   Undersampling removes majority class rows, which in this dataset would mean
   discarding tens of thousands of legitimate customer interactions. With already-
   small minority classes (Score 2 has only 1,283 rows), undersampling the
   majority would leave a very small training set, risking underfitting.

3. Applied to training set only:
   SMOTE was deliberately applied only to X_train, never to X_test. The test
   set must reflect the real-world class distribution to give honest, realistic
   performance metrics. Oversampling the test set would artificially inflate
   performance on minority classes and give a misleading picture of how the
   model would perform in production.

````python
# ML Model - 3 Implementation with Hyperparameter Optimization (RandomizedSearchCV)
# FIXED VERSION — handles numpy array output from SMOTE correctly

from sklearn.model_selection import RandomizedSearchCV
from xgboost import XGBClassifier
from sklearn.metrics import (accuracy_score, f1_score, precision_score,
                              recall_score, classification_report)
import numpy as np

# ── Convert to numpy if not already (handles both DataFrame and array) ───────
X_train_bal_np = X_train_bal if isinstance(X_train_bal, np.ndarray) else X_train_bal.values
y_train_xgb_np = y_train_xgb if isinstance(y_train_xgb, np.ndarray) else np.array(y_train_xgb)

print(f"X_train_bal type : {type(X_train_bal_np)}, shape: {X_train_bal_np.shape}")
print(f"y_train_xgb type : {type(y_train_xgb_np)}, shape: {y_train_xgb_np.shape}")

# ── Sample 20k rows for tuning ───────────────────────────────────────────────
sample_size = 20000
idx    = np.random.choice(len(X_train_bal_np), size=sample_size, replace=False)
X_tune = X_train_bal_np[idx]   # numpy fancy indexing — works correctly
y_tune = y_train_xgb_np[idx]

print(f"\nTuning sample — X: {X_tune.shape}, y: {y_tune.shape}")

# ── Parameter grid ───────────────────────────────────────────────────────────
param_dist_xgb = {
    'n_estimators'    : [100, 200],
    'max_depth'       : [3, 5, 7],
    'learning_rate'   : [0.05, 0.1, 0.2],
    'subsample'       : [0.8, 1.0],
    'colsample_bytree': [0.8, 1.0],
    'gamma'           : [0, 0.1]
}

random_xgb = RandomizedSearchCV(
    XGBClassifier(random_state=42, use_label_encoder=False,
                  eval_metric='mlogloss', n_jobs=-1, tree_method='hist'),
    param_distributions=param_dist_xgb,
    n_iter=10,
    cv=3,
    scoring='f1_macro',
    random_state=42,
    n_jobs=-1,
    verbose=1
)

# ── Fit on sample ────────────────────────────────────────────────────────────
print("\n⏳ Tuning on 20k sample — should finish in 5-10 minutes...")
random_xgb.fit(X_tune, y_tune)
print("✅ Tuning complete!")
print("Best Parameters :", random_xgb.best_params_)
print("Best CV Macro F1:", round(random_xgb.best_score_, 4))

# ── Retrain best model on FULL training data ─────────────────────────────────
print("\n⏳ Retraining best model on full training data...")
best_xgb = XGBClassifier(
    **random_xgb.best_params_,
    random_state=42,
    use_label_encoder=False,
    eval_metric='mlogloss',
    n_jobs=-1,
    tree_method='hist'
)
best_xgb.fit(X_train_bal_np, y_train_xgb_np)
print("✅ Final model trained on full data!")

# ── Predict (map 0-4 back to 1-5) ───────────────────────────────────────────
X_test_np        = X_test_scaled if isinstance(X_test_scaled, np.ndarray) else X_test_scaled.values
y_pred_xgb_tuned = best_xgb.predict(X_test_np) + 1

print("\n=== XGBoost (Tuned) — Evaluation ===")
print(f"Accuracy  : {accuracy_score(y_test, y_pred_xgb_tuned):.4f}")
print(f"Macro F1  : {f1_score(y_test, y_pred_xgb_tuned, average='macro'):.4f}")
print(f"Precision : {precision_score(y_test, y_pred_xgb_tuned, average='macro'):.4f}")
print(f"Recall    : {recall_score(y_test, y_pred_xgb_tuned, average='macro'):.4f}")
print(classification_report(y_test, y_pred_xgb_tuned))
````

##### Which hyperparameter optimization technique have you used and why?

RandomizedSearchCV with 5-fold cross-validation was used for XGBoost as well.

Why RandomizedSearchCV for XGBoost specifically:
XGBoost has more hyperparameters than Random Forest (learning_rate, max_depth,
subsample, colsample_bytree, gamma, n_estimators, etc.), creating an even larger
search space. GridSearchCV is completely infeasible here — 4×4×4×3×3×3 = 1,728
combinations × 5 folds = 8,640 model fits. RandomizedSearchCV with n_iter=20
covers 20 × 5 = 100 fits, finding a near-optimal solution at ~1% of the
computational cost. The key parameters to tune in XGBoost are learning_rate
(step size — smaller values need more trees), max_depth (tree complexity),
subsample and colsample_bytree (stochastic regularization), and gamma
(minimum loss reduction to make a further partition), all included in the search.

### 1. Which Evaluation metrics did you consider for a positive business impact and why?

Two metrics were prioritized above the others:

1. Macro F1 Score — Primary metric:
   Macro F1 averages the F1 score of each CSAT class equally, regardless of how
   many samples each class has. This is the most important metric for this problem
   because the business objective is to identify dissatisfied customers (the
   minority classes — CSAT 1 and 2), not just the majority (CSAT 5). A model
   that only performs well on the majority class would show high accuracy but low
   Macro F1 — Macro F1 catches this failure.

2. Recall (Macro) — Secondary metric:
   From a business perspective, the cost of missing a genuinely dissatisfied
   customer (false negative) is higher than the cost of incorrectly flagging a
   satisfied customer for follow-up (false positive). A missed dissatisfied
   customer may churn, leave a negative review, or escalate publicly. A falsely
   flagged satisfied customer just receives an unnecessary follow-up call —
   a minor inefficiency. High recall on minority classes minimizes the
   higher-cost error.

### 1. CSAT Distribution is Highly Imbalanced and Polarized
69.4% of customers gave a score of 5 (very satisfied), while only 1.49% gave a 2

handled using SMOTE during model training.

### 2. Response Time is the Single Strongest Driver of Dissatisfaction
Satisfied customers (CSAT 4-5) had a median response time of just 5 minutes,
while dissatisfied customers (CSAT 1-2) waited a median of 21 minutes — 4.2x
longer. This relationship was statistically confirmed (Mann-Whitney U, p ≈ 0.00),
and response_time_log was the most important feature in the final XGBoost model
by a significant margin.

### 3. Support Channel Significantly Affects Satisfaction (p = 1.53e-33)
Email had a dissatisfaction rate of 23.16%, compared to 14.32% for Inbound and
13.95% for Outcall. Post-hoc testing showed Email is significantly worse than
both other channels, while Inbound and Outcall are statistically indistinguishable
(p = 0.085). Email's failure is concentrated in CSAT 1 (21% of Email tickets),
pointing to a specific, fixable service failure mode rather than general weakness.

### 4. Agent Tenure Significantly Affects Satisfaction (p = 3.38e-47)
On Job Training agents had the highest dissatisfaction rate (16.64%), while the
61-90 day tenure bucket performed best (12.16% dissatisfaction). Satisfaction
improves consistently with experience up to ~61-90 days, then slightly declines
for the >90 day group, suggesting a potential burnout or complacency effect in
the most experienced cohort. Given that ~30% of all tickets are handled by
agents still in training, onboarding quality is a high-leverage intervention point.

### 5. Cancellation Category is a Disproportionate Pain Point
Despite not being the highest-volume category, Cancellations had a 21.47%
dissatisfaction rate — significantly worse than Returns (12.18%), which dominates
volume. Customers reaching out about cancellations are already frustrated before
the interaction begins, making the support experience at this touchpoint critically
important.

### 6. Morning Shift Bears Disproportionate Load with Worst Outcomes
The Morning shift handles 48.2% of all ticket volume (41,426 tickets) while having
the highest dissatisfaction rate (15.75%) — compared to the Split shift, which has
the lowest volume and the best dissatisfaction rate (10.33%). The heavy load
concentration in a single shift is likely suppressing service quality.

### 7. Reverse Pickup Enquiry Dominates Support Volume
A single sub-category — Reverse Pickup Enquiry — accounts for ~26% of all tickets
(22,389). Any improvement to the reverse pickup experience (better tracking,
proactive SMS updates, self-service options) would have an outsized operational
impact given this concentration.

   model doesn't fully leverage.

   further.
